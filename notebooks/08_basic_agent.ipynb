```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_number(a:int,b:int):
    """
    it will return the sum of two number
    Args:
        a (int): first number
        b(int): second number
    """
    return a+b

@tool
def sub_number(a:int,b:int):
    """
    it will return the subtraction of two number
    Args:
        a (int): first number
        b(int): second number
    """
    return a-b
```

```python
add_number.invoke({"a":13,"b":2})
sub_number.invoke({"a":13,"b":2})
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import create_agent
```

```python
llm=ChatGoogleGenerativeAI(model="gemini-2.5-flash")
```

```python
agent=create_agent(
    model=llm,
    tools=[add_number,sub_number],
    system_prompt="You are a math teacher and always use a tool for a calculation",
    
)
```

```python
res=agent.invoke({"messages":[{"role":"user","content":"what is 13+2*10"}]})
print(res["messages"][-1].content)
```
