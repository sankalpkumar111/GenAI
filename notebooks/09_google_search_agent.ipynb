```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain.agents import create_agent
from langchain_groq import ChatGroq
from langgraph.checkpoint.memory import MemorySaver
```

```python
llm=ChatGroq(model="openai/gpt-oss-20b")
search = GoogleSerperAPIWrapper()
```

```python
search.run("What is the capital of France?")
```

```python
agent=create_agent(
    model=llm,
    tools=[search.run],
    system_prompt="You are a helpful assistant that can answer questions using the Google Search API. Use the search tool to find information and provide accurate answers to the user's queries.",
    checkpointer=MemorySaver()
)
```

```python
question = "What is the Predisent"
```

```python
res = agent.invoke({"messages":[{"role":"user","content":question}]}, 
                 {"configurable": {"thread_id": "sankalp"}})
```

```python
print(res["messages"][-1].content)
```
